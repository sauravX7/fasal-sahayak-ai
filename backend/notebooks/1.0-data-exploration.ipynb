```python
# Cell 1: Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cell 2: Load the Data
# Make sure to replace 'your_dataset.csv' with the actual filename
df = pd.read_csv('../data/raw/your_dataset.csv')
print(df.head())
print(df.info())

# Cell 3: Filter for our Target Area and Crop
# Let's focus on Wheat in our home state, Madhya Pradesh
mp_wheat = df[(df['State_Name'] == 'Madhya Pradesh') & (df['Crop'] == 'Wheat')]
print(mp_wheat.head())

# Cell 4: Basic Analysis - Calculate Yield
# Yield is a crucial metric: Production / Area
# We need to handle cases where Area is 0 to avoid division errors
mp_wheat['Yield'] = mp_wheat['Production'] / mp_wheat['Area']
print(mp_wheat.describe())
```
